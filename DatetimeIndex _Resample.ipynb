### DatetimeIndex and Resample

```python
import pandas as pd

df = pd.read_csv(
    "aapl.csv",
    parse_dates=["Date"],
    index_col="Date",
    date_format="%d-%b-%y"
)

df.head(5)
```

```python
df.index
```

```python
df.loc['7-Jul-17']
```

```python
df.loc['2017-01']
```

```python
df.loc['2017-06'].head()
```

```python
df.loc["2017-06"].Close.mean()
```

```python
df.loc['2017'].head(2) 
```

```python
df = df.sort_index()
df.loc['2017-06-01':'2017-06-30']

```

```python
df['Close'].resample('ME').mean().head()
```

```python
%matplotlib inline
df['Close'].plot()
```
