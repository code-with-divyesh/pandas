### Diffrent way to create a DataFrame

```python
import pandas as pd
```

```python
#using csv file
df=pd.read_csv('weather_data.csv')
df
```

```python
#using excel file
df=pd.read_excel("weather_data.xlsx","Sheet1",engine="openpyxl")
df
```

```python
#using Dictionary
data = {
    'day': ['1/1/2017', '1/2/2017', '1/3/2017'],
    'temperature': [32, 35, 28],
    'windspeed': [6, 7, 2],
    'event': ['Rain', 'Sunny', 'Snow']
}

df = pd.DataFrame(data)
df
```

```python
# using List of tuple
data = [
    ('1/1/2017', 32, 6, 'Rain'),
    ('1/2/2017', 35, 7, 'Sunny'),
    ('1/3/2017', 28, 2, 'Snow')
]

df


```

```python
#List of Dictionary

data = [
    {'day': '1/1/2017', 'temperature': 32, 'windspeed': 6, 'event': 'Rain'},
    {'day': '1/2/2017', 'temperature': 35, 'windspeed': 7, 'event': 'Sunny'},
    {'day': '1/3/2017', 'temperature': 28, 'windspeed': 2, 'event': 'Snow'}
]

df = pd.DataFrame(data)
print(df)
```
