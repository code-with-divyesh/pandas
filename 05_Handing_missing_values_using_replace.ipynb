### Handing missing values using replace

```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv('replace.csv')
df
```

```python
#replace value based on specific column
new_df=df.replace({
  'temperature':-99999,
  'windspeed':[-99999,-88888],
  'event':'0',
},np.nan)
new_df
```

```python
new_df=df.replace({
  -99999:np.nan,
  'no event':'sunny'
})
new_df
```

```python

# when windspeed is 6 mph, 7 mph etc. & temperature is 32 F, 28 F etc.
new_df = df.replace({'temperature': '[A-Za-z]', 'windspeed': '[a-z]'},'', regex=True) 
new_df
```
