## Handling missing value 

```python
import pandas as pd
```

```python
df=pd.read_csv('weather_data_missing_values.csv')
df
```

```python
type(df.day[0])
```

```python
#convert day type ->str-time stamp
df=pd.read_csv('weather_data_missing_values.csv',parse_dates=['day'])
df.set_index('day',inplace=True)
df
```

```python
new_df=df.fillna(0)
new_df
```

```python
# file nan by specific column releted values
new_df=df.fillna({
  'temperature':0,
  'windspeed':0,
  'event':'no event'
})
new_df
```

```python
# handling missing value with forward fill
new_df=df.ffill()
new_df
```

```python
# handling missing value with forward fill
new_df=df.ffill(limit=1)
new_df
```

```python
# handling missing value with Backward fill
new_df=df.bfill()
new_df
```

```python
# handling missing value using axis
new_df=df.ffill(axis=0)
new_df
```

```python
# handling missing value with nearest one and approx 
new_df=df.interpolate()
new_df
```

```python
# handling missing value with nearest one and approx with time
new_df=df.interpolate(method='time')
new_df
```

```python
# Rows delete where value is nan
new_df=df.dropna()
new_df
```

```python
# Rows delete where value is nan in all 
new_df=df.dropna(how="all")
new_df
```

```python
# Rows delete where value is nan in all 
new_df=df.dropna(thresh=2)
new_df
```
